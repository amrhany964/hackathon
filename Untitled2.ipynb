```python
def parse_obj(x):
    if isinstance(x, (list, dict)):
        return x

    if x is None:
        return None

    try:
        if pd.isna(x):
            return None
    except:
        pass

    x = str(x).strip()

    try:
        return json.loads(x)
    except:
        pass

    try:
        return ast.literal_eval(x)
    except:
        pass

    if "," in x:
        return [item.strip() for item in x.split(",")]

    return [x]
```

```python
def clean_text(text):
    text = str(text)

    text = text.replace("ـ", "")
    text = re.sub(r"[\u064B-\u065F]", "", text)

    text = re.sub("[إأآا]", "ا", text)
    text = re.sub("ى", "ي", text)
    text = re.sub("ؤ", "و", text)
    text = re.sub("ئ", "ي", text)

    text = re.sub(r"http\S+|www\S+", " رابط ", text)
    text = re.sub(r"@\w+", " مستخدم ", text)

    # Keep Arabic, English, digits
    text = re.sub(r"[^\w\s\u0600-\u06FF]", " ", text)

    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
def prepare_aspect_data(df):
    rows = []

    for _, row in df.iterrows():
        aspects = parse_obj(row["aspects"])

        if aspects is None:
            aspects = ["none"]

        if isinstance(aspects, str):
            aspects = [aspects]

        aspects = [str(a).strip() for a in aspects]
        aspects = [a for a in aspects if a in ASPECTS_ALL]

        if len(aspects) == 0:
            aspects = ["none"]

        rows.append({
            "review_id": row["review_id"],
            "review_text": clean_text(row["review_text"]),
            "star_rating": row["star_rating"] if "star_rating" in df.columns else "",
            "platform": row["platform"] if "platform" in df.columns else "",
            "business_category": row["business_category"] if "business_category" in df.columns else "",
            "aspects_list": aspects
        })

    return pd.DataFrame(rows)


def build_aspect_input(row):
    return (
        f"التقييم: {row.get('star_rating', '')} [SEP] "
        f"المنصة: {row.get('platform', '')} [SEP] "
        f"النشاط: {row.get('business_category', '')} [SEP] "
        f"النص: {row['review_text']}"
    )
```

```python
aspect_train_raw = train.drop_duplicates(
    subset=["review_text", "aspects"]
).reset_index(drop=True)

aspect_valid_raw = validation.copy()

aspect_train_df = prepare_aspect_data(aspect_train_raw)
aspect_valid_df = prepare_aspect_data(aspect_valid_raw)

aspect_train_df["aspect_input"] = aspect_train_df.apply(build_aspect_input, axis=1)
aspect_valid_df["aspect_input"] = aspect_valid_df.apply(build_aspect_input, axis=1)

print(aspect_train_df.shape)
print(aspect_valid_df.shape)
display(aspect_train_df.head())
```

```python
mlb = MultiLabelBinarizer(classes=ASPECTS_ALL)

y_aspect_train = mlb.fit_transform(aspect_train_df["aspects_list"])
y_aspect_valid = mlb.transform(aspect_valid_df["aspects_list"])

print("Aspect classes:", list(mlb.classes_))
print("Train label matrix:", y_aspect_train.shape)
print("Valid label matrix:", y_aspect_valid.shape)
```

```python
aspect_model = Pipeline([
    ("tfidf", TfidfVectorizer(
        analyzer="char_wb",
        ngram_range=(3, 5),
        max_features=150000,
        min_df=2,
        sublinear_tf=True
    )),
    ("clf", OneVsRestClassifier(
        LogisticRegression(
            C=3.0,
            max_iter=3000,
            class_weight="balanced",
            solver="liblinear",
            random_state=SEED
        )
    ))
])

aspect_model.fit(aspect_train_df["aspect_input"], y_aspect_train)

print("Aspect model trained.")
```

```python
valid_aspect_probs = aspect_model.predict_proba(aspect_valid_df["aspect_input"])

for threshold in [0.15, 0.20, 0.25, 0.30, 0.35, 0.40, 0.45, 0.50]:
    valid_aspect_pred = (valid_aspect_probs >= threshold).astype(int)

    micro = f1_score(
        y_aspect_valid,
        valid_aspect_pred,
        average="micro",
        zero_division=0
    )

    macro = f1_score(
        y_aspect_valid,
        valid_aspect_pred,
        average="macro",
        zero_division=0
    )

    print(f"threshold={threshold:.2f} | micro-F1={micro:.4f} | macro-F1={macro:.4f}")
```

```python
valid_aspect_probs = aspect_model.predict_proba(aspect_valid_df["aspect_input"])

for threshold in [0.15, 0.20, 0.25, 0.30, 0.35, 0.40, 0.45, 0.50]:
    valid_aspect_pred = (valid_aspect_probs >= threshold).astype(int)

    micro = f1_score(
        y_aspect_valid,
        valid_aspect_pred,
        average="micro",
        zero_division=0
    )

    macro = f1_score(
        y_aspect_valid,
        valid_aspect_pred,
        average="macro",
        zero_division=0
    )

    print(f"threshold={threshold:.2f} | micro-F1={micro:.4f} | macro-F1={macro:.4f}")
```

```python
aspect_thresholds = {
    "food": 0.35,
    "service": 0.35,
    "price": 0.30,
    "cleanliness": 0.25,
    "delivery": 0.25,
    "ambiance": 0.30,
    "app_experience": 0.25,
    "general": 0.35,
    "none": 0.50
}
```

```python
def predict_aspects_from_probs(probs, threshold_dict):
    predicted_aspects_all = []

    for prob_row in probs:
        prob_dict = {
            aspect: prob_row[i]
            for i, aspect in enumerate(mlb.classes_)
        }

        predicted = []

        for aspect in ASPECTS_NO_NONE:
            threshold = threshold_dict.get(aspect, 0.35)

            if prob_dict.get(aspect, 0) >= threshold:
                predicted.append(aspect)

        if len(predicted) == 0:
            none_score = prob_dict.get("none", 0)

            if none_score >= threshold_dict.get("none", 0.50):
                predicted = ["none"]
            else:
                best_aspect = max(
                    ASPECTS_NO_NONE,
                    key=lambda a: prob_dict.get(a, 0)
                )
                predicted = [best_aspect]

        predicted_aspects_all.append(predicted)

    return predicted_aspects_all
```

```python
valid_pred_aspects = predict_aspects_from_probs(
    valid_aspect_probs,
    aspect_thresholds
)

y_valid_pred_aspects = mlb.transform(valid_pred_aspects)

print("Aspect micro-F1:", f1_score(
    y_aspect_valid,
    y_valid_pred_aspects,
    average="micro",
    zero_division=0
))

print("Aspect macro-F1:", f1_score(
    y_aspect_valid,
    y_valid_pred_aspects,
    average="macro",
    zero_division=0
))

print(classification_report(
    y_aspect_valid,
    y_valid_pred_aspects,
    target_names=mlb.classes_,
    zero_division=0
))
```

```python
def explode_train_for_sentiment(df):
    rows = []

    for _, row in df.iterrows():
        review_id = row["review_id"]
        review_text = str(row["review_text"])

        aspects = parse_obj(row["aspects"])
        aspect_sentiments = parse_obj(row["aspect_sentiments"])

        if aspects is None:
            continue

        if isinstance(aspects, str):
            aspects = [aspects]

        if not isinstance(aspect_sentiments, dict):
            continue

        for aspect in aspects:
            aspect = str(aspect).strip()

            if aspect == "none":
                continue

            if aspect not in ASPECTS_NO_NONE:
                continue

            sentiment = aspect_sentiments.get(aspect)

            if sentiment not in SENTIMENTS:
                continue

            rows.append({
                "review_id": review_id,
                "review_text": review_text,
                "star_rating": row["star_rating"] if "star_rating" in df.columns else "",
                "platform": row["platform"] if "platform" in df.columns else "",
                "business_category": row["business_category"] if "business_category" in df.columns else "",
                "aspect": aspect,
                "sentiment": sentiment
            })

    return pd.DataFrame(rows)


def build_input(row):
    aspect_name = aspect_ar.get(row["aspect"], row["aspect"])

    return (
        f"التقييم: {row.get('star_rating', '')} [SEP] "
        f"المنصة: {row.get('platform', '')} [SEP] "
        f"النشاط: {row.get('business_category', '')} [SEP] "
        f"الجانب: {aspect_name} [SEP] "
        f"النص: {row['review_text']}"
    )
```

```python
def balance_train_dataset(
    train_df,
    group_cols=["aspect", "sentiment"],
    min_samples_per_group=40,
    max_aug_per_group=60,
    replace_prob=0.35,
    random_state=42
):
    random.seed(random_state)

    synonym_dict = {
        "ممتاز": ["رائع", "جميل جدا", "حلو جدا"],
        "ممتازة": ["رائعة", "جميلة جدا", "حلوة جدا"],
        "جميل": ["حلو", "رائع", "ممتاز"],
        "جميلة": ["حلوة", "رائعة", "ممتازة"],
        "حلو": ["جميل", "كويس", "رائع"],
        "حلوة": ["جميلة", "كويسة", "رائعة"],
        "كويس": ["جيد", "حلو", "مقبول جدا"],
        "كويسة": ["جيدة", "حلوة", "مقبولة جدا"],
        "جيد": ["كويس", "مقبول", "لطيف"],
        "رائع": ["ممتاز", "جميل جدا", "تحفة"],
        "تحفة": ["رائع", "ممتاز", "جميل جدا"],
        "لذيذ": ["شهي", "طعمه جميل", "حلو"],
        "نظيف": ["مرتب", "نظيف جدا"],
        "سريع": ["سريع جدا", "في الوقت", "بدون تأخير"],

        "سيء": ["وحش", "رديء", "غير جيد"],
        "سيئ": ["وحش", "رديء", "غير جيد"],
        "سئ": ["وحش", "رديء", "غير جيد"],
        "وحش": ["سيء", "رديء", "غير مقبول"],
        "رديء": ["سيء", "وحش", "غير جيد"],
        "غالي": ["مرتفع السعر", "مبالغ فيه", "سعره عالي"],
        "بطيء": ["متأخر", "بطئ جدا", "فيه تأخير"],
        "بطئ": ["متأخر", "بطيء جدا", "فيه تأخير"],
        "متأخر": ["بطيء", "متأخر جدا", "فيه تأخير"],
        "وسخ": ["غير نظيف", "قذر", "مش نظيف"],
        "مزعج": ["غير مريح", "سيء", "متعب"],

        "عادي": ["مقبول", "متوسط", "لا بأس به"],
        "عادية": ["مقبولة", "متوسطة", "لا بأس بها"],
        "مقبول": ["عادي", "متوسط", "لا بأس به"],
        "متوسط": ["عادي", "مقبول"],

        "good": ["great", "nice", "decent"],
        "great": ["excellent", "amazing", "wonderful"],
        "nice": ["good", "pleasant", "lovely"],
        "love": ["adore", "really like", "enjoy"],
        "bad": ["poor", "terrible", "awful"],
        "slow": ["delayed", "very slow", "late"],
        "expensive": ["overpriced", "costly", "too expensive"],
        "dirty": ["unclean", "not clean", "filthy"],
        "ok": ["average", "acceptable", "fine"],
        "average": ["acceptable", "okay", "normal"]
    }

    protected_words = {
        "مش", "مو", "لا", "ما", "لم", "لن", "ليس", "ليست", "بدون", "غير",
        "not", "no", "never", "n't",
        "but", "لكن", "بس", "ولكن"
    }

    def synonym_replace_text(text):
        words = str(text).split()
        new_words = []
        changed = False

        for word in words:
            clean_word = word.strip(".,!?؟،؛:()[]{}\"'").lower()

            if clean_word in protected_words:
                new_words.append(word)
                continue

            if clean_word in synonym_dict and random.random() < replace_prob:
                new_words.append(random.choice(synonym_dict[clean_word]))
                changed = True
            else:
                new_words.append(word)

        return " ".join(new_words), changed

    df = train_df.copy()
    df["is_augmented"] = 0

    augmented_rows = []

    print("Before augmentation:")
    print(df.groupby(group_cols).size().sort_values())

    for group_key, group in df.groupby(group_cols):
        n = len(group)

        if n >= min_samples_per_group:
            continue

        needed = min(min_samples_per_group - n, max_aug_per_group)

        sampled = group.sample(
            n=needed,
            replace=True,
            random_state=random_state
        )

        for _, row in sampled.iterrows():
            original_text = str(row["review_text"])
            new_text = original_text
            changed = False

            for _ in range(8):
                candidate_text, changed = synonym_replace_text(original_text)
                if changed and candidate_text != original_text:
                    new_text = candidate_text
                    break

            if not changed:
                continue

            new_row = row.copy()
            new_row["review_text"] = new_text
            new_row["is_augmented"] = 1
            new_row["input_text"] = build_input(new_row)

            augmented_rows.append(new_row)

    if len(augmented_rows) > 0:
        aug_df = pd.DataFrame(augmented_rows)
        final_df = pd.concat([df, aug_df], ignore_index=True)
    else:
        final_df = df.copy()

    final_df = final_df.sample(frac=1, random_state=random_state).reset_index(drop=True)

    print("\nAugmented rows added:", len(augmented_rows))
    print("\nAfter augmentation:")
    print(final_df.groupby(group_cols).size().sort_values())
    print("\nOld shape:", train_df.shape)
    print("New shape:", final_df.shape)

    return final_df
```

```python
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=1)

    return {
        "macro_f1": f1_score(labels, preds, average="macro"),
        "accuracy": accuracy_score(labels, preds)
    }
```

```python
class SimpleTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        labels = inputs.get("labels")
        outputs = model(**inputs)
        logits = outputs.get("logits")

        loss_fct = nn.CrossEntropyLoss()
        loss = loss_fct(logits.view(-1, 3), labels.view(-1))

        return (loss, outputs) if return_outputs else loss
```

```python
def train_transformer_model(
    model_name,
    train_df,
    valid_df,
    output_dir,
    epochs=4,
    lr=2e-5,
    batch_size=8,
    max_length=160
):
    tokenizer = AutoTokenizer.from_pretrained(model_name)

    train_ds = Dataset.from_pandas(
        train_df[["input_text", "label"]].reset_index(drop=True)
    )

    valid_ds = Dataset.from_pandas(
        valid_df[["input_text", "label"]].reset_index(drop=True)
    )

    def tokenize_batch(batch):
        return tokenizer(
            batch["input_text"],
            truncation=True,
            max_length=max_length
        )

    train_ds = train_ds.map(tokenize_batch, batched=True)
    valid_ds = valid_ds.map(tokenize_batch, batched=True)

    train_ds = train_ds.rename_column("label", "labels")
    valid_ds = valid_ds.rename_column("label", "labels")

    collator = DataCollatorWithPadding(tokenizer=tokenizer)

    model = AutoModelForSequenceClassification.from_pretrained(
        model_name,
        num_labels=3,
        id2label=id2sentiment,
        label2id=sentiment2id
    )

    args = TrainingArguments(
        output_dir=output_dir,
        learning_rate=lr,
        per_device_train_batch_size=batch_size,
        per_device_eval_batch_size=batch_size * 2,
        num_train_epochs=epochs,
        weight_decay=0.01,
        logging_steps=50,
        save_total_limit=1,
        report_to="none",
        fp16=False,
        seed=SEED
    )

    trainer = SimpleTrainer(
        model=model,
        args=args,
        train_dataset=train_ds,
        eval_dataset=valid_ds,
        data_collator=collator,
        compute_metrics=compute_metrics
    )

    trainer.train()

    return trainer, tokenizer
```

```python
def predict_proba_transformer(trainer, tokenizer, texts, max_length=160):
    ds = Dataset.from_dict({
        "input_text": list(texts)
    })

    def tokenize_batch(batch):
        return tokenizer(
            batch["input_text"],
            truncation=True,
            max_length=max_length
        )

    ds = ds.map(tokenize_batch, batched=True)

    preds = trainer.predict(ds)
    logits = preds.predictions

    logits = logits - logits.max(axis=1, keepdims=True)
    exp_logits = np.exp(logits)
    probs = exp_logits / exp_logits.sum(axis=1, keepdims=True)

    return probs
```

```python
trainer_xlm, tokenizer_xlm = train_transformer_model(
    model_name=DOMAIN_MODEL_DIR,
    train_df=train_df,
    valid_df=valid_df,
    output_dir="./models/sentiment_domain_xlm",
    epochs=4,
    lr=2e-5,
    batch_size=8,
    max_length=160
)

valid_probs_xlm = predict_proba_transformer(
    trainer_xlm,
    tokenizer_xlm,
    valid_df["input_text"]
)

valid_pred_xlm = valid_probs_xlm.argmax(axis=1)

f1_xlm = f1_score(valid_df["label"], valid_pred_xlm, average="macro")

print("Domain XLM-R Macro F1:", f1_xlm)
print(classification_report(valid_df["label"], valid_pred_xlm, target_names=SENTIMENTS))
```

```python
trainer_marbert, tokenizer_marbert = train_transformer_model(
    model_name=MARBERT_MODEL,
    train_df=train_df,
    valid_df=valid_df,
    output_dir="./models/sentiment_marbert",
    epochs=4,
    lr=2e-5,
    batch_size=8,
    max_length=160
)

valid_probs_marbert = predict_proba_transformer(
    trainer_marbert,
    tokenizer_marbert,
    valid_df["input_text"]
)

valid_pred_marbert = valid_probs_marbert.argmax(axis=1)

f1_marbert = f1_score(valid_df["label"], valid_pred_marbert, average="macro")

print("MARBERT Macro F1:", f1_marbert)
print(classification_report(valid_df["label"], valid_pred_marbert, target_names=SENTIMENTS))
```

```python
trainer_camel, tokenizer_camel = train_transformer_model(
    model_name=CAMEL_MODEL,
    train_df=train_df,
    valid_df=valid_df,
    output_dir="./models/sentiment_camelbert",
    epochs=4,
    lr=2e-5,
    batch_size=8,
    max_length=160
)

valid_probs_camel = predict_proba_transformer(
    trainer_camel,
    tokenizer_camel,
    valid_df["input_text"]
)

valid_pred_camel = valid_probs_camel.argmax(axis=1)

f1_camel = f1_score(valid_df["label"], valid_pred_camel, average="macro")

print("CAMeLBERT Macro F1:", f1_camel)
print(classification_report(valid_df["label"], valid_pred_camel, target_names=SENTIMENTS))
```

```python
weight_grid = [
    (0.40, 0.40, 0.20),
    (0.45, 0.40, 0.15),
    (0.50, 0.35, 0.15),
    (0.35, 0.45, 0.20),
    (0.33, 0.33, 0.34),
    (0.50, 0.25, 0.25),
    (0.25, 0.50, 0.25),
    (0.25, 0.25, 0.50),
]

best_f1 = -1
best_weights = None
best_valid_probs = None
best_valid_pred = None

for w in weight_grid:
    probs = (
        w[0] * valid_probs_xlm +
        w[1] * valid_probs_marbert +
        w[2] * valid_probs_camel
    )

    pred = probs.argmax(axis=1)
    score = f1_score(valid_df["label"], pred, average="macro")

    print("Weights:", w, "Macro F1:", score)

    if score > best_f1:
        best_f1 = score
        best_weights = w
        best_valid_probs = probs
        best_valid_pred = pred

print("\nBest weights:", best_weights)
print("Best ensemble F1:", best_f1)
print(classification_report(valid_df["label"], best_valid_pred, target_names=SENTIMENTS))
```

```python
def build_submission_with_predicted_aspects(test_original, test_aspect_pred_df, test_sentiment_df):
    sentiment_map = {}

    if test_sentiment_df is not None and len(test_sentiment_df) > 0:
        for _, row in test_sentiment_df.iterrows():
            review_id = int(row["review_id"])
            aspect = str(row["aspect"]).strip()
            sentiment = str(row["predicted_sentiment"]).strip()

            if review_id not in sentiment_map:
                sentiment_map[review_id] = {}

            sentiment_map[review_id][aspect] = sentiment

    aspect_map = {}

    for _, row in test_aspect_pred_df.iterrows():
        aspect_map[int(row["review_id"])] = row["predicted_aspects"]

    submission = []

    for _, row in test_original.iterrows():
        review_id = int(row["review_id"])
        aspects = aspect_map.get(review_id, ["none"])

        if aspects is None:
            aspects = ["none"]

        if isinstance(aspects, str):
            aspects = [aspects]

        if aspects == ["none"] or len(aspects) == 0:
            submission.append({
                "review_id": review_id,
                "aspects": ["none"],
                "aspect_sentiments": {
                    "none": "neutral"
                }
            })
            continue

        clean_aspects = []
        aspect_sentiments = {}

        for aspect in aspects:
            aspect = str(aspect).strip()

            if aspect == "none":
                continue

            if aspect not in ASPECTS_NO_NONE:
                continue

            sentiment = sentiment_map.get(review_id, {}).get(aspect, "neutral")

            clean_aspects.append(aspect)
            aspect_sentiments[aspect] = sentiment

        if len(clean_aspects) == 0:
            submission.append({
                "review_id": review_id,
                "aspects": ["none"],
                "aspect_sentiments": {
                    "none": "neutral"
                }
            })
        else:
            submission.append({
                "review_id": review_id,
                "aspects": clean_aspects,
                "aspect_sentiments": aspect_sentiments
            })

    return submission


submission = build_submission_with_predicted_aspects(
    test,
    test_aspect_df,
    test_df
)

display(submission[:5])
```

```python
def validate_submission(submission):
    allowed_aspects = [
        "food",
        "service",
        "price",
        "cleanliness",
        "delivery",
        "ambiance",
        "app_experience",
        "general",
        "none"
    ]

    allowed_sentiments = ["positive", "negative", "neutral"]

    assert isinstance(submission, list), "Submission must be a list"

    seen_ids = set()

    for i, item in enumerate(submission):
        assert isinstance(item, dict), f"Item {i} must be a dict"

        assert "review_id" in item, f"Missing review_id at item {i}"
        assert "aspects" in item, f"Missing aspects at item {i}"
        assert "aspect_sentiments" in item, f"Missing aspect_sentiments at item {i}"

        assert isinstance(item["review_id"], int), f"review_id must be int at item {i}"
        assert isinstance(item["aspects"], list), f"aspects must be list at item {i}"
        assert isinstance(item["aspect_sentiments"], dict), f"aspect_sentiments must be dict at item {i}"

        assert item["review_id"] not in seen_ids, f"Duplicate review_id {item['review_id']}"
        seen_ids.add(item["review_id"])

        assert len(item["aspects"]) == len(set(item["aspects"])), f"Duplicate aspects at item {i}"

        for aspect in item["aspects"]:
            assert aspect in allowed_aspects, f"Invalid aspect {aspect} at item {i}"
            assert aspect in item["aspect_sentiments"], f"Missing sentiment for {aspect} at item {i}"
            assert item["aspect_sentiments"][aspect] in allowed_sentiments, f"Invalid sentiment at item {i}"

        for aspect in item["aspect_sentiments"]:
            assert aspect in item["aspects"], f"{aspect} in sentiment dict but not aspects at item {i}"

    print("Submission is valid.")
    print("Rows:", len(submission))


validate_submission(submission)
```
